```python
import numpy as np
from sklearn.metrics import confusion_matrix, accuracy_score, recall_score, precision_score, f1_score, mean_absolute_error, mean_squared_error, r2_score
```

# Metryki klasyfikacji
# 1. y_test - rzeczywiste numery klas
# 2. y_pred - przewidziane przez model numery klas

```python
y_test_k = np.random.default_rng(seed=42).integers(low=0, high=5, size=10000)
y_pred_k = [y_test_k[idx] if x == 'y_test' else np.random.choice(np.delete([0, 1, 2, 3, 4], np.where(np.array([0, 1, 2, 3, 4]) == y_test_k[idx])[0]))
          for idx, x in enumerate(np.random.default_rng(seed=42).choice(['y_test', 'other_class'], size=10000, p=[0.85, 0.15]))]
```

## Confusion Matrix (Macierz Pomyłek)

### Własna funkcja

```python
def ConfusionMatrix(y_test, y_pred):
    return np.array(
        [
            [
                [
                    1 if pred == col and test == row else 0 for test, pred in zip(y_test, y_pred)
                ].count(1)
                for col in range(0, len(np.unique(y_test)))
            ]
            for row in range(0, len(np.unique(y_test)))
        ]
    )
```

```python
print(ConfusionMatrix(y_test_k, y_pred_k))
```

### Wbudowana funkcja

```python
print(confusion_matrix(y_test_k, y_pred_k))
```

### Accuracy (Dokładność)

### Własna funkcja

```python
def Accuracy(y_test, y_pred):
    return [1 if y_test[idx] == y_pred[idx] else 0 for idx in range(0, len(y_test))].count(1) / len(y_test)
```

```python
print(Accuracy(y_test_k, y_pred_k))
print(Accuracy(y_test_k, y_pred_k)*100)
```

### Wbudowana funkcja

```python
print(accuracy_score(y_test_k, y_pred_k))
print(accuracy_score(y_test_k, y_pred_k)*100)
```

### Recall (Czułość)

### Własna funkcja

```python
def Recall(y_test, y_pred):
    matrix = ConfusionMatrix(y_test, y_pred)
    
    return [
        round(float(matrix[idx][idx] / sum(matrix[idx])), 8)
        for idx in range(0, len(np.unique(y_test)))
    ]
```

```python
print(Recall(y_test_k, y_pred_k))
```

### Wbudowana funkcja

```python
print(recall_score(y_test_k, y_pred_k, average=None))
```

### Precision (Precyzja)

### Własna funkcja

```python
def Precision(y_test, y_pred):
    matrix = ConfusionMatrix(y_test, y_pred)
    
    return [
        round(float(matrix[idx][idx] / sum([row[idx:idx+1][0] for row in matrix])), 8)
        for idx in range(0, len(np.unique(y_test)))
    ]
```

```python
print(Precision(y_test_k, y_pred_k))
```

### Wbudowana funkcja

```python
print(precision_score(y_test_k, y_pred_k, average=None))
```

### F1-score

### Własna funkcja

```python
def F1(y_test, y_pred):
    return [
        round(2 * (precision * recall) / (precision + recall), 8)
        for precision, recall in zip(Precision(y_test, y_pred), Recall(y_test, y_pred))
    ]
```

```python
print(F1(y_test_k, y_pred_k))
```

### Wbudowana funkcja

```python
print(f1_score(y_test_k, y_pred_k, average=None))
```

# Metryki regresji

## 1. y_test - rzeczywiste wartości
## 2. y_pred - przewidziane wartości przez model

```python
y_test_r = np.random.default_rng(seed=42).uniform(low=100, high=2000, size=10000)
y_pred_r = np.random.default_rng(seed=42).uniform(low=50, high=2500, size=10000)
```

### MAE (Mean Absolute Error)

### Własna funkcja

```python
def MAE(y_test, y_pred):
    return sum([abs(test - pred) for test, pred in zip(y_test, y_pred)]) / len(y_test)
```

```python
print(MAE(y_test_r, y_pred_r))
```

### Wbudowana funkcja

```python
print(mean_absolute_error(y_test_r, y_pred_r))
```

### MSE (Mean Squared Error)

### Własna funkcja

```python
def MSE(y_test, y_pred):
    return sum([(test - pred)**2 for test, pred in zip(y_test, y_pred)]) / len(y_test)
```

```python
print(MSE(y_test_r, y_pred_r))
```

### Wbudowana funkcja

```python
print(mean_squared_error(y_test_r, y_pred_r))
```

### RMSE (Root Mean Squared Error)

### Własna funkcja

```python
def RMSE(y_test, y_pred):
    return np.sqrt(MSE(y_test, y_pred))
```

```python
print(RMSE(y_test_r, y_pred_r))
```

### Wbudowana funkcja

```python
print(np.sqrt(mean_squared_error(y_test_r, y_pred_r)))
```

### R^2 (Współczynnik determinacji)

### Własna funkcja

```python
def R2(y_test, y_pred):
    y_test_mean = sum(y_test) / len(y_test)
    return 1 - sum([(test - pred)**2 for test, pred in zip(y_test, y_pred)]) / sum([(test - y_test_mean)**2 for test in y_test])
```

```python
print(R2(y_test_r, y_pred_r))
```

### Wbudowana funkcja

```python
print(r2_score(y_test_r, y_pred_r))
```

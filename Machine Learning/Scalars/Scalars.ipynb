```python
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

# Sztuczne dane do testów

```python
X1 = np.random.default_rng(seed=42).integers(low=1, high=100, size=10000).tolist()
X2 = np.random.default_rng(seed=42).integers(low=1000, high=100000, size=10000).tolist()
X3 = np.random.default_rng(seed=42).integers(low=0, high=9, size=10000).tolist()
X = np.array(list(zip(X1, X2, X3)))

X_train = X[:8000, :]
X_test = X[8000:]

print('X_train:\n', X_train)
print('\nX_test:\n', X_test)
```

# StandardScaler

## Wbudowana funkcja

```python
scaler = StandardScaler()

X_train_sc = scaler.fit_transform(X_train)
X_test_sc = scaler.transform(X_test)

print('X_train (StandardScaler):\n', X_train_sc)
print('\nX_test (StandardScaler):\n', X_test_sc)
```

## Własna funkcja

```python
class standard_scaler:
    def __init__(self):
        self.X_train = None
        self.X_test = None
        self.mean = None
        self.std = None
        
    def FitTransform(self, X):
        self.X_train = X
        self.mean = [sum(self.X_train[:, idx]) / len(self.X_train[:, idx]) for idx in range(0, self.X_train.shape[1])]
        self.std = [np.sqrt(sum((self.X_train[:, idx] - self.mean[idx])**2) / self.X_train.shape[0]) for idx in range(0, self.X_train.shape[1])]

        return np.array(
            [
                [
                    (self.X_train[row, col] - self.mean[col]) / self.std[col]
                    for col in range(0, self.X_train.shape[1])
                ]
                for row in range(0, self.X_train.shape[0])
            ]
        )

    def Transform(self, X):
        self.X_test = X

        return np.array(
            [
                [
                    (self.X_test[row, col] - self.mean[col]) / self.std[col]
                    for col in range(0, self.X_test.shape[1])
                ]
                for row in range(0, self.X_test.shape[0])
            ]
        )
```

```python
scaler2 = standard_scaler()

X_train_sc2 = scaler2.FitTransform(X_train)
X_test_sc2 = scaler2.Transform(X_test)

print('X_train (Class: standard_scaler):\n', X_train_sc2)
print('\nX_test (Class: standard_scaler):\n', X_test_sc2)
```

# MinMaxScaler

## Wbudowana funkcja

```python
scaler3 = MinMaxScaler()

X_train_min_max_sc = scaler3.fit_transform(X_train)
X_test_min_max_sc = scaler3.transform(X_test)

print('X_train (MinMaxScaler):\n', X_train_min_max_sc)
print('\nX_test (MinMaxScaler):\n', X_test_min_max_sc)
```

## Własna funkcja

```python
class min_max_scaler:
    def __init__(self):
        self.X_train = None
        self.X_test = None
        self.max = None
        self.min = None

    def FitTransform(self, X):
        self.X_train = X
        self.max = [np.max(self.X_train[:, idx]) for idx in range(0, self.X_train.shape[1])]
        self.min = [np.min(self.X_train[:, idx]) for idx in range(0, self.X_train.shape[1])]

        return np.array(
            [
                [
                    (self.X_train[row, col] - self.min[col]) / (self.max[col] - self.min[col])
                    for col in range(0, self.X_train.shape[1])
                ]
                for row in range(0, self.X_train.shape[0])
            ]
        )

    def Transform(self, X):
        self.X_test = X

        return np.array(
            [
                [
                    (self.X_test[row, col] - self.min[col]) / (self.max[col] - self.min[col])
                    for col in range(0, self.X_test.shape[1])
                ]
                for row in range(0, self.X_test.shape[0])
            ]
        )
```

```python
scaler4 = min_max_scaler()

X_train_min_max_sc2 = scaler4.FitTransform(X_train)
X_test_min_max_sc2 = scaler4.Transform(X_test)

print('X_train (Class: min_max_scaler):\n', X_train_min_max_sc2)
print('\nX_test (Class: min_max_scaler):\n', X_test_min_max_sc2)
```
